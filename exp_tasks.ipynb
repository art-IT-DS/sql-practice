```python
sqldf("SELECT order_status, COUNT(*) AS count FROM orders GROUP BY order_status ORDER BY count DESC")
```

```python
sqldf("""
    SELECT 
        order_status,
        COUNT(*) AS count, 
        MIN(order_purchase_timestamp) AS first_order, 
        MAX(order_purchase_timestamp) AS last_order
    FROM orders
    GROUP BY order_status
    ORDER BY count DESC
""")
```

```python
sqldf("""
    SELECT order_status, COUNT(*) AS count
    FROM orders
    GROUP BY order_status
    HAVING COUNT(*) > 1000
""")
```

```python
sqldf("""
    SELECT order_status, COUNT(*) AS count
    FROM orders
    WHERE order_purchase_timestamp >= '2018-01-01'
    GROUP BY order_status
    HAVING COUNT(*) > 100
    ORDER BY count DESC
""")
```

```python
sqldf("""
    SELECT
        orders.order_id,
        orders.order_status,
        customers.customer_city,
        customers.customer_state
    FROM orders
    JOIN customers ON orders.customer_id = customers.customer_id
    LIMIT 10
""")
```

```python
sqldf("""
    SELECT
        orders.order_id,
        customers.customer_city,
        order_items.price,
        order_items.freight_value
    FROM orders
    JOIN customers ON orders.customer_id = customers.customer_id
    JOIN order_items ON orders.order_id = order_items.order_id
    LIMIT 10
""")
```
